# Feature Creation

## `date_ist` — calendar date in IST

```python
df['date_ist'] = df['time_ist'].dt.date
df['date_ist'].head()
```

## `year_month` — year and month as a string (e.g. 2020-07)

```python
df['year_month'] = df['time_ist'].dt.to_period('M').astype(str)
df['year_month'].value_counts().sort_index()
```

## `month_num` — month as a number (1 = January … 12 = December)

```python
df['month_num'] = df['time_ist'].dt.month
df['month_num'].value_counts().sort_index()
```

## `month_name` — month as a full name (e.g. July)

```python
df['month_name'] = df['time_ist'].dt.strftime('%B')
df['month_name'].value_counts()
```

## `day_of_month` — day number within the month (1 to 31)

```python
df['day_of_month'] = df['time_ist'].dt.day
df['day_of_month'].describe()
```

## `weekday_num` — day of week as a number (Monday = 1, Sunday = 7)

```python
df['weekday_num'] = df['time_ist'].dt.weekday + 1
df['weekday_num'].value_counts().sort_index()
```

## `weekday_name` — day of week as a name (e.g. Monday)

```python
df['weekday_name'] = df['time_ist'].dt.strftime('%A')
df['weekday_name'].value_counts()
```

## `is_weekend` — True if Saturday or Sunday

```python
df['is_weekend'] = df['weekday_num'].isin([6, 7])
df['is_weekend'].value_counts()
```

## `hour_ist` — hour of the tweet in IST (0 to 23)

```python
df['hour_ist'] = df['time_ist'].dt.hour
df['hour_ist'].value_counts().sort_index()
```

## `word_count` — number of words in the tweet

```python
df['word_count'] = df['Tweet'].str.split().str.len()
df['word_count'].describe()
```

## `char_count` — number of characters in the tweet

```python
df['char_count'] = df['Tweet'].str.len()
df['char_count'].describe()
```

## `has_D_upper` — True if tweet contains uppercase letter D (case-sensitive)

```python
df['has_D_upper'] = df['Tweet'].str.contains('D', regex=False)
df['has_D_upper'].value_counts()
```

## `has_H_upper` — True if tweet contains uppercase letter H (case-sensitive)

```python
df['has_H_upper'] = df['Tweet'].str.contains('H', regex=False)
df['has_H_upper'].value_counts()
```

## `has_media` — True if media views > 0

```python
df['has_media'] = df['media views'] > 0
df['has_media'].value_counts()
```

## `has_link` — True if url clicks > 0

```python
df['has_link'] = df['url clicks'] > 0
df['has_link'].value_counts()
```

## `has_hashtag` — True if hashtag clicks > 0

```python
df['has_hashtag'] = df['hashtag clicks'] > 0
df['has_hashtag'].value_counts()
```

## `has_app_open` — True if app opens > 0

```python
df['has_app_open'] = df['app opens'] > 0
df['has_app_open'].value_counts()
```

## `media_label` — readable label for has_media

```python
df['media_label'] = df['has_media'].map({True: 'With Media', False: 'Without Media'})
df['media_label'].value_counts()
```

## `app_open_label` — readable label for has_app_open

```python
df['app_open_label'] = df['has_app_open'].map({True: 'With App Opens', False: 'Without App Opens'})
df['app_open_label'].value_counts()
```

## `category` — Media first, then Links, then Hashtags, else Text Only

```python
df['category'] = 'Text Only'
df.loc[df['has_hashtag'], 'category'] = 'Hashtags'
df.loc[df['has_link'],    'category'] = 'Links'
df.loc[df['has_media'],   'category'] = 'Media'
df['category'].value_counts()
```

## `rt_plus_likes` — sum of retweets and likes

```python
df['rt_plus_likes'] = df['retweets'] + df['likes']
df['rt_plus_likes'].describe()
```

# Data Analysis

Which month had the highest total impressions?

```python
df.groupby('year_month')['impressions'].sum().sort_index()
```

July 2020 had by far the most impressions (345,286), nearly double any other month.

Which tweet category has the highest average engagement rate?

```python
df.groupby('category')['engagement rate'].mean().round(4).sort_values(ascending=False)
```

Link tweets have the highest avg engagement rate (5.97%), followed by Media (5.15%) — Text Only is lowest (2.34%).

Which day of the week gets the most retweets?

```python
df.groupby('weekday_name')['retweets'].sum().sort_values(ascending=False)
```

Friday dominates with 202 total retweets — nearly 4x more than any other day.

Do weekend tweets get a higher engagement rate than weekday tweets?

```python
df.groupby('is_weekend')['engagement rate'].mean().round(4)
```

Weekend tweets have a slightly higher avg engagement rate (4.24%) than weekday tweets (3.55%).

What are the top 10 tweets by combined retweets and likes?

```python
df.nlargest(10, 'rt_plus_likes')[['Tweet', 'rt_plus_likes', 'retweets', 'likes']]
```

The top tweet has 2,652 rt_plus_likes (165 retweets + 2,487 likes), far ahead of all others.

Do tweets with media get a higher engagement rate than tweets without media?

```python
df.groupby('media_label')['engagement rate'].mean().round(4)
```

Tweets with media have an avg engagement rate of 5.15% vs 2.74% without media — nearly double.

At what hour (IST) are tweets most commonly posted?

```python
df.groupby('hour_ist')['id'].count().sort_index()
```

Most tweets are posted late night (11 PM = 118 tweets) and midnight (12 AM = 99 tweets) IST. Hours 9-12 have almost no tweets.

Which category drives the most total user interactions (url + profile + hashtag clicks)?

```python
df['total_clicks'] = df['url clicks'] + df['user profile clicks'] + df['hashtag clicks']
df.groupby('category')['total_clicks'].sum().sort_values(ascending=False)
```

Media tweets generate the most clicks (2,603), followed by Links (1,937) — Hashtag tweets are lowest despite having a dedicated click type.

How does the average engagement rate change across months?

```python
df.groupby('year_month')['engagement rate'].mean().round(4).sort_index()
```

Engagement rate is fairly stable across months (3.4%-3.9%), with September and October slightly higher.

Is there a difference in average word count across tweet categories?

```python
df.groupby('category')['word_count'].mean().round(1).sort_values(ascending=False)
```

All categories have nearly identical average word counts (~26 words), so tweet length does not explain engagement differences.

# Filter Check

## Filter 1 — Tweet Interaction Breakdown: even day, word_count > 40, any click > 0

```python
f1 = df[
    (df['day_of_month'] % 2 == 0) &
    (df['word_count'] > 40) &
    (df['total_clicks'] > 0)
]
print('Rows matching Filter 1:', len(f1))
print('Note: max word_count in dataset is', df['word_count'].max(), '— filter requires > 40, so 0 rows match.')
```

## Filter 2 — Engagement Rate Comparison: weekday, hour 9-16, even impressions, odd day, char > 30, no uppercase D

```python
f2 = df[
    (~df['is_weekend']) &
    (df['hour_ist'] >= 9) & (df['hour_ist'] <= 16) &
    (df['impressions'] % 2 == 0) &
    (df['day_of_month'] % 2 != 0) &
    (df['char_count'] > 30) &
    (~df['has_D_upper'])
]
print('Rows matching Filter 2:', len(f2))
```

## Filter 3 — Media Interaction by Day: months 7-9, even impressions, odd day, char > 30, no uppercase H

```python
f3 = df[
    (df['month_num'] >= 7) & (df['month_num'] <= 9) &
    (df['impressions'] % 2 == 0) &
    (df['day_of_month'] % 2 != 0) &
    (df['char_count'] > 30) &
    (~df['has_H_upper'])
]
print('Rows matching Filter 3:', len(f3))
```

## Filter 4 — Replies, Retweets, Likes: months 6-8 (June, July, August)

```python
f4 = df[
    (df['month_num'] >= 6) & (df['month_num'] <= 8)
]
print('Rows matching Filter 4:', len(f4))
```

## Filter 5 — Top 10 Tweets: weekday, even impressions, odd day, word_count < 30

```python
f5 = df[
    (~df['is_weekend']) &
    (df['impressions'] % 2 == 0) &
    (df['day_of_month'] % 2 != 0) &
    (df['word_count'] < 30)
]
print('Rows matching Filter 5:', len(f5))
```
